```python
import os
from PIL import Image

import numpy as np
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from sklearn.preprocessing import StandardScaler
```

```python
def load_images(folder_path):
    X = []
    y = []

    class_names = ["fresh", "moldy"]

    for label, class_name in enumerate(class_names):
        class_path = os.path.join(folder_path, class_name)

        for img_name in os.listdir(class_path):
            img_path = os.path.join(class_path, img_name)

            img = Image.open(img_path).convert("RGB")
            img = img.resize((IMG_SIZE, IMG_SIZE))

            img = np.array(img) / 255.0
            X.append(img.flatten())
            y.append(label)

    return np.array(X), np.array(y)
```

```python
import os

IMG_SIZE = 64
DATASET_DIR = r"C:\Users\rimuk\Downloads\Bread_Mold_Datasets\bread_dataset"

print(os.listdir(DATASET_DIR))

```

```python
X_train, y_train = load_images(os.path.join(DATASET_DIR, "train"))
X_val, y_val     = load_images(os.path.join(DATASET_DIR, "val"))
X_test, y_test   = load_images(os.path.join(DATASET_DIR, "test"))
```

```python
print(X_train.shape, X_val.shape, X_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_val   = scaler.transform(X_val)
X_test  = scaler.transform(X_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(
    n_neighbors=5,
    metric="euclidean"
)

knn.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score

val_pred = knn.predict(X_val)
print("KNN Validation Accuracy:", accuracy_score(y_val, val_pred))
```

```python
from sklearn.metrics import confusion_matrix, classification_report

test_pred = knn.predict(X_test)

print("KNN Test Accuracy:", accuracy_score(y_test, test_pred))
print(confusion_matrix(y_test, test_pred))
print(classification_report(y_test, test_pred, target_names=["Fresh", "Moldy"]))
```

```python
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
svm = LinearSVC(C=1.0, max_iter=5000)

svm.fit(X_train, y_train)
```

```python
val_pred_svm = svm.predict(X_val)
print("SVM Validation Accuracy:", accuracy_score(y_val, val_pred_svm))
```

```python
test_pred_svm = svm.predict(X_test)

print("SVM Test Accuracy:", accuracy_score(y_test, test_pred_svm))
print(confusion_matrix(y_test, test_pred_svm))
print(classification_report(y_test, test_pred_svm, target_names=["Fresh", "Moldy"]))
```
